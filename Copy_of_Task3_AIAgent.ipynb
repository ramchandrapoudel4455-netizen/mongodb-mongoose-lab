```

# Student Support AI Agent
# CN6005 Artificial Intelligence Coursework

print("===================================")
print("     STUDENT SUPPORT AI AGENT")
print("===================================")

print("Welcome! I am your Student Support Assistant.")
print("I can help answer your university questions.")

```

```

# Creating a simple student support knowledge base

knowledge_base = {
    "coursework": "Students can find coursework instructions and deadlines on their university learning platform.",

    "library": "Students can use the university library to access books, computers and study spaces.",

    "student_support": "Students can contact the student support team for help with academic or personal difficulties.",

    "registration": "Students should contact the university administration team for enrolment and registration enquiries.",

    "timetable": "Students can check their university timetable through their student portal."
}

print("Knowledge base created successfully!")
print("Number of topics:", len(knowledge_base))
```

```

# Function to search the knowledge base

def search_knowledge_base(question):

    question = question.lower()

    for topic, answer in knowledge_base.items():

        if topic.replace("_", " ") in question:
            return answer

    return "Sorry, I could not find information about your question."

# Testing the search function

question = "Where can I find coursework information?"

answer = search_knowledge_base(question)

print("Student Question:", question)
print("Assistant Answer:", answer)
```

```

# Interactive Student Support Assistant

print("Welcome to Student Support AI!")
print("Type 'exit' to stop the assistant.")

while True:

    question = input("\nAsk your question: ")

    if question.lower() == "exit":
        print("Thank you for using Student Support AI!")
        break

    answer = search_knowledge_base(question)

    print("Assistant:", answer)
```

```
def generate_ai_response(question):
    information = search_knowledge_base(question)

    prompt = f"""
    You are a friendly Student Support AI Assistant.

    Use the following information to answer the student's question.

    Knowledge base: {information}

    Student question: {question}

    Answer in simple and clear English.
    If the information is not available, do not invent an answer.
    """

    response = client.interactions.create(
        model="gemini-3.1-flash-lite",
        input=prompt
    )

    return response.output_text

print("AI response function created successfully!")
```
